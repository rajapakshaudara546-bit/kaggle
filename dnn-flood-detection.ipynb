```python
import yaml
import os

TABULAR_FOLDER = "/kaggle/input/datasets/uom230429e/sri-lanka-flood-tabular-graph-2003-2025"
IMAGE_FOLDER = "/kaggle/input/datasets/uom230429e/flood-data-set" 

# 1. Automatically hunt for the 'frames' folder
correct_sar_root = None
for root, dirs, files in os.walk(IMAGE_FOLDER):
    if 'frames' in dirs:
        correct_sar_root = root  # We found the parent of 'frames'!
        break

if correct_sar_root:
    print(f"🎉 FOUND SAR FRAMES AT: {correct_sar_root}/frames")
    
    # 2. Update config
    config_path = "configs/kaggle_data.yaml"
    with open(config_path, "r") as f:
        config = yaml.safe_load(f)
        
    config["data_paths"]["panel"] = f"{TABULAR_FOLDER}/flood_dataset.parquet"
    config["data_paths"]["nodes"] = f"{TABULAR_FOLDER}/nodes.csv"
    config["data_paths"]["edges_flow"] = f"{TABULAR_FOLDER}/edges.csv"
    config["data_paths"]["edges_spatial"] = f"{TABULAR_FOLDER}/edges.csv"
    
    # Put the automatically found path here
    config["data_paths"]["sar_chips"] = correct_sar_root
    
    with open(config_path, "w") as f:
        yaml.dump(config, f)
    
    print("✅ Config updated! You can start training now.")
else:
    print(f"❌ COULD NOT FIND 'frames' FOLDER ANYWHERE IN {IMAGE_FOLDER}!")
```
